```python
import numpy as np

# 뉴런의 출력 계산 함수
def calculate(input):
    global weights
    global bias
    activation = bias		# 바이어스
    for i in range(2):		# 입력 신호 총합 계산
        activation += weights[i] * input[i]
    if round(activation, 10) > 0.0:	# 계단 함수
        return 1.0 
    else:
        return 0.0
 
# 학습 알고리즘
def train_weights(X, y, l_rate, n_epoch):
    global weights
    global bias
    for epoch in range(n_epoch):		# 에포크 반복
        sum_error = 0.0
        for row, target in zip(X, y):		# 데이터셋을 반복
            actual = calculate(row)		# 실제 출력 계산
            error = target - actual		# 오차 계산
            bias = bias + l_rate * error	
            sum_error += error**2		# 오차의 제곱 계산
            for i in range(2):			# 가중치 변경
                weights[i] = weights[i] + l_rate * error * row[i]
            print("입력:", row, "목표:", target, "실제출력:", actual, "오차:", error, "변경된 가중치와 바이어스:", [round(i, 2) for i in weights], round(bias, 2))
        print('에포크 번호=%d, 학습률=%.2f, 오차=%d개' % (epoch+1, l_rate, sum_error))
        print()
        if sum_error == 0:
          break    
    return weights


# <학습데이터의 입력과 정답> 입력
X = [[160,60],[163,60],[160,64],[163,64],[165,61]]
y = [0, 0, 1, 1, 1]

x1=np.array(X)
min_value = np.min(x1)
max_value = np.max(x1) 
X=(x1-0)/(max_value-0)
print(X)


# 가중치와 바이어스 초기값 설정
weights = [0.4, -0.1]
bias = 0.1
l_rate = 0.1		# 학습률설정

n_epoch = 1000		# 최대 에포크 횟수 임의 설정
weights = train_weights(X, y, l_rate, n_epoch)
print()
print("최종 가중치와 바이어스")
print([round(i, 2) for i in weights], round(bias, 2))
```

```python
# 예측 수행하기(테스트)

test = [162, 61]
test=np.array(test)
test=(test-0)/(max_value-0)

predict = calculate(test)		# 실제 출력 계산
print(predict)
```

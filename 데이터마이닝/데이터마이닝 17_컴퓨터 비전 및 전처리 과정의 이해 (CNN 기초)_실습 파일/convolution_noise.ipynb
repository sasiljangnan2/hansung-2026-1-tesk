```python
# OpenCV import 및 원본 영상 출력

import numpy as np
import cv2
from pathlib import Path

try:
    from google.colab.patches import cv2_imshow
except Exception:
    import matplotlib.pyplot as plt

    def cv2_imshow(image):
        if image is None:
            raise ValueError("표시할 이미지가 없습니다.")
        if image.ndim == 3 and image.shape[2] == 3:
            image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        plt.imshow(image, cmap="gray" if image.ndim == 2 else None)
        plt.axis("off")
        plt.show()

candidate_paths = [
    Path("lena256_rgb_noise.jpg"),
    Path.cwd() / "lena256_rgb_noise.jpg",
    Path("/content/lena256_rgb_noise.jpg"),
    Path(r"c:/Users/cth62/OneDrive/바탕 화면/과제/깃/hansung-2026-1-tesk/lena256_rgb_noise.jpg"),
    Path(r"c:/Users/cth62/OneDrive/바탕 화면/과제/깃/hansung-2026-1-tesk/데이터마이닝/lena256_rgb_noise.jpg"),
    Path(r"c:/Users/cth62/OneDrive/바탕 화면/과제/깃/hansung-2026-1-tesk/데이터마이닝/데이터마이닝 17_컴퓨터 비전 및 전처리 과정의 이해 (CNN 기초)_실습 파일/lena256_rgb_noise.jpg"),
]

img_path = next((p for p in candidate_paths if p.exists()), None)

if img_path is None:
    try:
        from google.colab import files
        uploaded = files.upload()
        if uploaded:
            first_name = next(iter(uploaded.keys()))
            img_path = Path(first_name)
    except Exception:
        pass

if img_path is None:
    raise FileNotFoundError(
        f"lena256_rgb_noise.jpg 파일을 찾지 못했습니다. 현재 작업 폴더: {Path.cwd()}"
    )

img = cv2.imread(str(img_path))
if img is None:
    raise ValueError(f"이미지 로드 실패: {img_path}")

print("원본 영상")
print(f"사용 경로: {img_path}")
cv2_imshow(img)
```

```python
# 3*3 스무딩 필터

filter1=np.array([[1/9, 1/9, 1/9], [1/9, 1/9, 1/9],[1/9, 1/9, 1/9]])
im_fil1=cv2.filter2D(img, -1, filter1)
print("3*3 스무딩 필터")
cv2_imshow(im_fil1)
```

```python
# 4*4 스무딩 필터

filter2=np.array([[1/16, 1/16, 1/16, 1/16], [1/16, 1/16, 1/16, 1/16], [1/16, 1/16, 1/16, 1/16], [1/16, 1/16, 1/16, 1/16]])
im_fil2=cv2.filter2D(img, -1, filter2)
print("4*4 스무딩 필터")
cv2_imshow(im_fil2)
```

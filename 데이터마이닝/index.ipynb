```python
import numpy as np


# 1차원 배열의 정수 인덱싱 vs Boolean 인덱싱

data = np.array([10, 20, 30])

subset = data[0]  # 하나의 데이터 접근
print(subset)

subset = data[ [0, 2] ]   # 1차원의 경우 여러 개의 데이터에 접근할 때는 
print(subset)             # data[0, 2]처럼 콤마를 사용하면 다차원 인덱싱으로 해석 
                          # 여러 개를 선택하려면 리스트(또는 ndarray)를 사용 

subset = data[[True, False, True]]   # Boolean 인덱싱
print(subset)



print("------------------")

data = np.array([
    ['초', 'A', '10'],
    ['중', 'B', '20'],
    ['초', 'C', '30']
])

subset_elem = data[ [0, 2] ]   # 0번째 행, 2번째 행 선택
print(subset_elem)

subset_elem = data[ [True, False, True] ]   # True인 행 선택(0번째 행, 2번째 행 선택)
print(subset_elem)

print("------------------")

data = np.array([
    ['초', 'A', '10'],
    ['중', 'B', '20'],
    ['초', 'C', '30']
])


subset = data[:, [0, 2]]   # 0번째 열, 2번째 열 선택
print(subset)


subset = data[:, [True, False, True] ]   # True인 열 선택(0번째 열, 2번째 열 선택)
print(subset)


```
